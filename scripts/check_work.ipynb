```python
# notebook to check work
import pandas as pd
import json
import csv
import numpy as np
from pathlib import Path

# Path setup
PROJECT_ROOT = Path(__file__).parent.parent if '__file__' in dir() else Path().absolute().parent
DATA_DIR = PROJECT_ROOT / "data"
PROCESSED_DIR = DATA_DIR / "processed"
RAW_DIR = DATA_DIR / "raw"

print(f"Project root: {PROJECT_ROOT}")
```

```python
# load csv:
step1 = PROCESSED_DIR / "frge_adultadhd_step1_494.csv"
step2 = PROCESSED_DIR / "frge_adultadhd_step2_464.csv"
step3 = PROCESSED_DIR / "frge_adultadhd_step3_464.csv"

df1 = pd.read_csv(step1)
df2 = pd.read_csv(step2)
df3 = pd.read_csv(step3)

idstep1 = list(set(df1['ID']))
idstep2 = list(set(df2['ID']))
idstep3 = list(set(df3['ID']))

print(len(idstep1))
print(len(idstep2))
print(len(idstep3))

# check if all ids are the same in idstep2 and idstep3:
idstep2.sort()
idstep3.sort()
idstep2 == idstep3
```

```python
# diff between idstep1 and idstep3:
diff = list(set(idstep1) - set(idstep3))
print(len(diff))
# save to csv
with open(PROCESSED_DIR / 'tdiff.csv', 'w') as myfile:
    wr = csv.writer(myfile, quoting=csv.QUOTE_ALL)
    for val in set(diff):
        wr.writerow([val])
```

```python
raw_group = RAW_DIR / "survey_data.csv"
df = pd.read_csv(raw_group)
id_raw_group = list(df['PROLIFIC_PID'])
print(f'total IDs from raw group list: {len(id_raw_group)}')

raw_trials = RAW_DIR / "total_adultadhd_trials.csv"
df_trials = pd.read_csv(raw_trials)
id_raw_trials = list(set(df_trials['ID']))
print(f'total IDs from raw trial list: {len(id_raw_trials)}')

first_cut = PROCESSED_DIR / "477_group_frge_09.19.21.csv"
df_first_cut = pd.read_csv(first_cut)
id_first_cut = list(set(df_first_cut['ID']))
print(f'total IDs from 092121 cut: {len(id_first_cut)}')

# take difference between raw_group and raw_trials:
bw2n1 = list(set(id_raw_trials)-set(id_raw_group))
bw2n3 = list(set(id_raw_trials)-set(id_first_cut))
bw1n3 = list(set(id_raw_group)-set(id_first_cut))
print(f'IDs in raw trials but not raw group: {len(bw2n1)}')
print(f'IDs in raw trials but not 092121 cut: {len(bw2n3)}')
print(f'IDs in 092121 cut but not raw group: {len(bw1n3)}')

# compare with idstep3:
bw3n4 = list(set(id_first_cut)-set(idstep3))
print(f'IDs in 092121 cut but not step3: {len(bw3n4)}')
# save bw3n4 to csv:
df_bw3n4 = pd.DataFrame(bw3n4)
df_bw3n4.to_csv(PROCESSED_DIR / 'bw3n4.csv')
```
