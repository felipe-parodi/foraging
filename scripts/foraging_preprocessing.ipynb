### Steps

### 2. Clean ID set.

```python
# Load data:
sorted_data = pd.read_csv('/Users/felipeparodi/Documents/code/foraging/fADHD_trials_022122.csv')
unsorted_data = pd.read_csv('/Users/felipeparodi/Documents/code/foraging/frge_adultadhd_step1.csv')

# Load ID columns:
sorted_data_IDs = set(sorted_data['ID'].tolist())
unsorted_data_IDs = set(unsorted_data['ID'].tolist())
leftover = unsorted_data_IDs - sorted_data_IDs

# Remove leftover IDs from unsorted data:
unsorted_data_with_newIDs = unsorted_data[~unsorted_data['ID'].isin(leftover)]
final_id_count = set(unsorted_data_with_newIDs['ID'].tolist())
print(f'Final count of ID list: ', len(final_id_count))

unsorted_data_with_newIDs.to_csv('/Users/felipeparodi/Documents/code/foraging/frge_adultadhd_step2.csv', index=False)
```

### 3. Remove invalid patches.

```python
sorted_data = pd.read_csv('/Users/felipeparodi/Documents/code/foraging/fADHD_trials_022122.csv')
unsorted_data2 = pd.read_csv('/Users/felipeparodi/Documents/code/foraging/frge_adultadhd_step2.csv')

print(f'First pass DF length: ', len(sorted_data))
print(f'Length before removing invalid patches: ', len(unsorted_data2))
unsorted_data2.head()

users = sorted(set(unsorted_data2['ID'].tolist()))
print(f'There are {len(users)} users.')

# Remove final bush of all users:
amp2 = pd.DataFrame([])

for i in range(len(users)):
    user = users[i]
    df = unsorted_data2[unsorted_data2['ID'] == user]
    
    df1 = df[df['Block']==1]
    bush1 = df1['Bush'].tolist()[-1]
    last_decision = df1['Decision'].tolist()[-1]
    if last_decision != 0:
        df1 = df1[df1['Bush'] != bush1]

    df2 = df[df['Block']==2]
    bush2 = df2['Bush'].tolist()[-1]
    last_decision = df2['Decision'].tolist()[-1]
    if last_decision != 0:
        df2 = df2[df2['Bush'] != bush2]

    df3 = df[df['Block']==3]
    bush3 = df3['Bush'].tolist()[-1]
    last_decision = df3['Decision'].tolist()[-1]
    if last_decision != 0:
        df3 = df3[df3['Bush'] != bush3]

    df4 = df[df['Block']==4]
    bush4 = df4['Bush'].tolist()[-1]
    last_decision = df4['Decision'].tolist()[-1]
    if last_decision != 0:
        df4 = df4[df4['Bush'] != bush4]

    # # Save to new dataframe:
    amp2 = amp2.append(df1)
    amp2 = amp2.append(df2)
    amp2 = amp2.append(df3)
    amp2 = amp2.append(df4)
print(f'Length after removing invalid bushes ', len(amp2))
amp2.to_csv('/Users/felipeparodi/Documents/code/foraging/frge_adultadhd_step3.csv')
```
