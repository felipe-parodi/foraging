```python
sorted_data = pd.read_csv(PROCESSED_DIR / 'fADHD_trials_022122.csv')
unsorted_data2 = pd.read_csv(PROCESSED_DIR / 'frge_adultadhd_step2.csv')

print(f'First pass DF length: ', len(sorted_data))
print(f'Length before removing invalid patches: ', len(unsorted_data2))
unsorted_data2.head()

users = sorted(set(unsorted_data2['ID'].tolist()))
print(f'There are {len(users)} users.')

# Remove final bush of all users IF final decision is not 0:
dfs_to_concat = []

for i in range(len(users)):
    user = users[i]
    df = unsorted_data2[unsorted_data2['ID'] == user]
    
    df1 = df[df['Block']==1]
    bush1 = df1['Bush'].tolist()[-1]
    last_decision = df1['Decision'].tolist()[-1]
    if last_decision != 0:
        df1 = df1[df1['Bush'] != bush1]

    df2 = df[df['Block']==2]
    bush2 = df2['Bush'].tolist()[-1]
    last_decision = df2['Decision'].tolist()[-1]
    if last_decision != 0:
        df2 = df2[df2['Bush'] != bush2]

    df3 = df[df['Block']==3]
    bush3 = df3['Bush'].tolist()[-1]
    last_decision = df3['Decision'].tolist()[-1]
    if last_decision != 0:
        df3 = df3[df3['Bush'] != bush3]

    df4 = df[df['Block']==4]
    bush4 = df4['Bush'].tolist()[-1]
    last_decision = df4['Decision'].tolist()[-1]
    if last_decision != 0:
        df4 = df4[df4['Bush'] != bush4]

    # Save to list for concatenation:
    dfs_to_concat.extend([df1, df2, df3, df4])

amp2 = pd.concat(dfs_to_concat, ignore_index=True)
print(f'Length after removing invalid bushes ', len(amp2))
amp2.to_csv(PROCESSED_DIR / 'frge_adultadhd_step3.csv')
```
